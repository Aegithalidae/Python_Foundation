```python
# division_calculator_plus.py
# 这一版改进了错误处理，使得报错后仍能输入数字重新计算

print("Give me two numbers, and I'll divide them.") 
print("Enter 'q' to quit.")

while True:
    first_num = input("\nFirst number: ")
    if first_num == 'q':
        break

    second_num = input("\nSecond number: ")
    if second_num == 'q':
        break

    try:
        answer = int(first_num)/int(second_num)

    except ZeroDivisionError:
        print("You can't divide by zero!")
        while True:
            second_num = input("Enter a nonzero number: ")
            if second_num == 'q':
                break
            try:
                answer = int(first_num) / int(second_num)
                print(answer)
                break
            except ZeroDivisionError:
                print("You can't divide by zero!")

    else:
        print(answer)
```

```python
# division_calculator_pro.py
# 之前的程序循环套循环，感觉很臃肿。
# 注意到这个简单的除法计算器只有一个功能和几个流程
# 所以写成函数是更好的选择

print("Give me two numbers, and I'll divide them.\nEnter 'q' to quit.") 

def division(num_1, num_2):
    try:
        answer = int(num_1)/int(num_2)
    except ZeroDivisionError:
        print("You can't divide by zero!")
    except ValueError:
        print("Please enter numbers only.")
    else:
        print(answer)
        return True
    

while True:
    first_num = input("\nFirst number: ")
    if first_num == 'q':
        break

    second_num = input("\nSecond number: ")
    if second_num == 'q':
        break

    division(first_num, second_num)
```

```python
def divide_numbers(num_1, num_2):
    """将两个字符串形式的数字转换为整数并相除。

    用户输入来自 input()，因此最初得到的是字符串。
    这个函数负责把输入转换为整数，执行除法，并处理常见异常。
    """
    try:
        answer = int(num_1) / int(num_2)
    except ZeroDivisionError:
        # 当第二个数字为 0 时，除法无法进行，但程序不应因此崩溃。
        print("You can't divide by zero!")
    except ValueError:
        # 当用户输入的内容无法转换为整数时，int() 会引发 ValueError。
        print("Please enter valid integers.")
    else:
        # 只有 try 代码块没有发生异常时，才打印计算结果。
        print(answer)


def main():
    """运行一个简单的交互式除法计算器。"""
    print("Give me two numbers, and I'll divide them.")
    print("Enter 'q' to quit.")

    while True:
        first_num = input("\nFirst number: ")
        if first_num == 'q':
            break

        second_num = input("Second number: ")
        if second_num == 'q':
            break

        divide_numbers(first_num, second_num)


if __name__ == "__main__":
    main()
```

```python
# 统计小说 Alice in Wonderland 的单词数，并处理文件不存在的情况。

def count_words(text):
    """统计文本中的单词数量。"""
    words = text.split()
    return len(words)


def main():
    """读取文本文件，并在文件存在时统计其中的单词数量。"""
    file_name = 'text_files/alice.txt'

    try:
        with open(file_name, encoding='utf-8') as file_object:
            contents = file_object.read()
    except FileNotFoundError:
        print(f"The file {file_name} can't be found!")
    else:
        word_count = count_words(contents)
        print(f"The file {file_name} has about {word_count} words.")


if __name__ == "__main__":
    main()
```

```python
# Jupyter 的所有单元格共享同一个命名空间；
# 因此如果 count_words() 已在前面的单元格中定义，这里可以直接调用。

novel_list = [
    'text_files/alice.txt',
    'text_files/siddhartha.txt',
    'text_files/little_women.txt',
    'text_files/moby_dick.txt',
]


def main():
    """批量读取小说文本，并统计每个文件的大致单词数。"""
    for novel in novel_list:
        try:
            with open(novel, encoding='utf-8') as novel_obj:
                contents = novel_obj.read()
        except FileNotFoundError:
            print(f"The file {novel} can't be found!")
        else:
            words = count_words(contents)
            print(f"The file {novel} has about {words} words.")


if __name__ == "__main__":
    main()
```

```python
# refactoring
import json
filename = 'json/username.json'

def check_json(filename):
    try:
        with open(filename):
            pass
    except FileNotFoundError:
        print("Sorry, the file does not exits\nIt'll be created soon")
        with open(filename, 'w'):
            pass
    else:
        print("The file is available now")

def remember_user(username):
    with open(filename, 'w') as file_obj:
        json.dump(username, file_obj)
        print(f"{username} has been reserved.")
        return username

def greet_user(name):
    print(f"Welcome, {name.title()}!")

def main():
    check_json(filename)
    name = input("Please enter your name: ")
    greet_user(remember_user(name))

if __name__ == "__main__":
    main()

'''只是把行为拆成了函数，但结构仍然很乱，没有体现到重构的整洁'''
```

```python
import json

filename = 'json/username.json'

def get_stored_username(filename):
    '''检查json是否存在，如果有姓名信息就获取它'''

    try:
        with open(filename, encoding='utf-8') as file_obj:
            username = json.load(file_obj)
    except FileNotFoundError:
        return None
    else:
        return username
    

def get_new_username(filename):
    '''向目标json中写入姓名，如果json不存在则创建后写入'''
    
    with open(filename, 'w') as file_obj:
        name = input("Please enter your name:")
        json.dump(name, file_obj)

    print(f"The {name} has been saved.")
    return name

def greet_user(filename):
    '''向目标json中的用户打招呼'''
    stored_username = get_stored_username(filename)
    if stored_username:
        print(f"Welcome back, {stored_username}")
    else:
        new_username = get_new_username(filename)
        print(f"Hello, {new_username}!")

def main():
    greet_user(filename)

if __name__ == "__main__":
    main()
```

```python
# 10-13 Verified User
import json

filename = 'json/username.json'
proper_name = 'wukaiyuan'

def check_file(file):
    '''检查名字json是否为空，若空返回None，非空则读取姓名并返回'''
    try:
        with open(file) as file_obj:
            stored_name = file_obj.read()
    except FileNotFoundError:
        return None
    else:
        return stored_name


def write_username(file):
    '''向目标文件写入指定姓名并保存，返回该姓名'''
    with open(file, 'w', encoding='utf-8') as file_obj:
        storing_name = input("Please enter your name: ")
        file_obj.write(storing_name)
    return storing_name.lower()


def check_username(name):
    '''检查用户名是否正确，返回布尔值'''
    if name != proper_name:
        return False
    elif name == proper_name:
        return True


def welcome_user(check_bool, name):
    if check_bool:
        print(f"Welcome, {name}")
    else:
        print(f"{name.title()}, you are not the proper user.")


def main():
    # 检查文件是否为存在，若不存在则提示创建
    if check_file(filename):
        name_read = check_file(filename)
    else:
        name_read = write_username(filename)

    # 检查姓名是否正确，若正确则欢迎，若不正确则提示错误
    if check_username(name_read):
        print(f"Welcome, {name_read}")
    else:
        name_again = input(f"{name_read.title()} is not the proper name, please again.").lower()
        while name_again != proper_name:
            name_again = input(f"{name_read.title()} is not the proper name, please again.").lower()

    # 打印欢迎词
    welcome_user(check_username(name_again), name_again)

if __name__ == "__main__":
    main()

'''
check_file 被调用了两次——第一次判断，第二次取值，文件被读取两遍，和之前 greet_user 里的重复调用问题一样。
check_username 的两个分支互斥，elif 可以直接写 else，或者更简洁地直接 return name == proper_name。
main() 末尾 welcome_user 引用了 name_again，但如果第一次 check_username(name_read) 就通过，name_again 根本没有被定义，会抛 NameError。
这是当前代码最严重的 bug。
另外这版放弃了 json，改用裸文本读写，与题目方向有些偏离。
'''
```

```python
import json

filename = 'json/username.json'
PROPER_NAME = 'wukaiyuan'

def check_file(filename):
    try:
        with open(filename) as file_obj:
            stored_name = json.load(file_obj)
    except FileNotFoundError:
        return None
    else:
        return stored_name

def write_file(filename, name):
    with open(filename, 'w', encoding='utf-8') as file_obj:
        json.dump(name, file_obj)


def main():
    name = check_file(filename)
    if name is None:
        name = input("Please enter your name: ").lower()
    
    while name != PROPER_NAME:
        print("It won't work until you enter the proper name.")
        name = input("Please enter the proper name: ").lower()
    
    print(f"Welcome, {name}")
    write_file(filename, name)


if __name__ == "__main__":
    main()
```
