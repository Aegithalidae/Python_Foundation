```python
import unittest
from survey import AnonymousSurvey

class TestAnonymousSurvey(unittest.TestCase):
    def test_store_single_response(self):
        '''测试单个答案会被保存'''
        question = "What language did you first learn to speak?"
        my_survey = AnonymousSurvey(question)
        my_survey.store_response('English')
        self.assertIn('English', my_survey.response)

    def test_store_single_responses(self):
        '''测试多个答案会被保存'''
        question = "What language did you first learn to speak?"
        my_survey = AnonymousSurvey(question)
        responses = ['English', 'Spanish', 'Mandarin']
        for response in responses:
            my_survey.store_response(response)
            self.assertIn(response, my_survey.response)

unittest.main(argv=[''], exit=False)
```

```python
import unittest
from survey import AnonymousSurvey


class TestAnonymousSurvey(unittest.TestCase):
    def setUp(self):
        """创建一个调查对象和一组答案，供测试使用。"""
        question = "What language did you first learn to speak?"
        self.my_survey = AnonymousSurvey(question)
        self.responses = ['English', 'Spanish', 'Mandarin']

    def test_store_single_response(self):
        """测试单个答案会被保存。"""
        self.my_survey.store_response(self.responses[0])
        self.assertIn(self.responses[0], self.my_survey.response)

    def test_store_multiple_responses(self):
        """测试多个答案会被保存。"""
        for response in self.responses:
            self.my_survey.store_response(response)

        for response in self.responses:
            self.assertIn(response, self.my_survey.response)


unittest.main(argv=[''], exit=False)
```

```python
import unittest
from employee import Employee

class TestEmployee(unittest.TestCase):
    def setUp(self):
        first_name = 'wu'
        last_name = 'kaiyuan'
        annual_salary = 100
        self.employee = Employee(first_name, last_name, annual_salary)

    def test_give_default_raise(self):
        self.employee.give_raise()
        self.assertEqual(int(5100), int(self.employee.annual_salary))

    def test_give_custom_raise(self):
        self.employee.give_raise(6000)
        self.assertEqual(int(6100), int(self.employee.annual_salary))

unittest.main(argv=[''], exit=False)
```
